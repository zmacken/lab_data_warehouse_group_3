```python
import os
import pandas as pd
import snowflake.connector
from dotenv import load_dotenv

load_dotenv()
conn = snowflake.connector.connect(
    user=os.getenv("SNOWFLAKE_USER"),
    password=os.getenv("SNOWFLAKE_PASSWORD"),
    account=os.getenv("SNOWFLAKE_ACCOUNT"),
    warehouse=os.getenv("SNOWFLAKE_WAREHOUSE"),
    database=os.getenv("SNOWFLAKE_DATABASE"),
    role=os.getenv("SNOWFLAKE_ROLE"),
)
df = conn.cursor().execute("select * from raw.flights").fetch_pandas_all()
df.columns = df.columns.str.lower()
```

```python
#lite hjälp från AI för att veta vad som är rätt sätt att koda en crosscheck emot CSV och df.
#saknade o göra en concat när jag testade utan hjälp...
sw_airports = pd.concat([df["flight_leg_identifier__departure_airport_iata"],
                         df["flight_leg_identifier__arrival_airport_iata"]]).unique()
missing_airports = sorted(set(sw_airports) - set(airports["IATA Code"].dropna()))
print(f"\nSwedavia-flygplatser utan träff i Wikipedia: {len(missing_airports)} av {len(sw_airports)}")
print(missing_airports)
```

```python
sw_airlines = df["airline_operator__icao"].unique()
missing_airlines = sorted(set(sw_airlines) - set(airlines["ICAO"].dropna()))
print(f"Swedavia-flygbolag (ICAO) utan träff i Wikipedia: {len(missing_airlines)} av {len(sw_airlines)}")
print(missing_airlines)
```

```python
dep, arr = "flight_leg_identifier__departure_airport_iata", "flight_leg_identifier__arrival_airport_iata"
wiki_iata = set(airports["IATA Code"].dropna())
```

```python
# vad AI kom ut med efter att jag försökte lista ut anledningen till "problemet" över.

# Andel flygrader där båda flygplatserna finns på Wikipedia
both = df[dep].isin(wiki_iata) & df[arr].isin(wiki_iata)
print(f"Rader där båda flygplatserna matchar: {both.mean():.1%}")
```
